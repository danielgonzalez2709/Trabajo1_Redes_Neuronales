## Generación de ciudades sintéticas

```python
# Generación de ciudades (sintéticas)
seed(56) # Para tener resultados reproducibles
n_ciudades = 40 # Número de ciudades
Ciudades = np.array(np.random.random(2*n_ciudades)).reshape(n_ciudades,2) # Ubicación de las ciudades
print(Ciudades)
```

## Cálculo de las distancias entre ciudades

```python
from scipy.spatial import distance_matrix
M = distance_matrix(Ciudades,Ciudades) # Matriz de distancias.
print(M)
```

# Visualización de las ciudades

```python
plt.scatter(Ciudades[:,0],Ciudades[:,1])
plt.title("Ciudades a visitar")
plt.grid()
plt.show()
```

# Optimización con Colonias de Hormigas

```python
# Se importa el optimizador que se usará
from AntColonyOptimizer import AntColonyOptimizer
```

```python
# Definción del problema e instaciación de la clase
problem = M
optimizer = AntColonyOptimizer(ants=40, evaporation_rate=.05, intensification=2, alpha=2, beta=1,
                               beta_evaporation_rate=0, choose_best=.1)
```

```python
# Optimización:
best = optimizer.fit(problem, 500)
# Visualización de la mejor solución:
optimizer.plot()
```

```python
# Visualización de la mejor solución
optimizer.best_path
```

```python
# Recorrido resultante con la mejor solución:
plt.scatter(Ciudades[:,0],Ciudades[:,1])
plt.plot(Ciudades[optimizer.best_path,0],Ciudades[optimizer.best_path,1],'-r')
plt.grid()
plt.title("Ciudades a visitar")
plt.show()
```
